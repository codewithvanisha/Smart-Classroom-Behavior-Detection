```python
import matplotlib.pyplot as plt
```

```python
behavior_counts = {
    "handrise": 1329,
    "read": 1,
    "sleep": 10,
    "write": 2
}
```

```python
labels = list(behavior_counts.keys())
values = list(behavior_counts.values())

plt.figure(figsize=(8,5))
plt.bar(labels, values)

plt.title("Classroom Behaviour Analysis")
plt.xlabel("Behaviour")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,6))
plt.pie(values, labels=labels, autopct='%1.1f%%')

plt.title("Behaviour Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(labels, values)
plt.title("Classroom Behaviour Analysis")
plt.xlabel("Behaviour")
plt.ylabel("Count")
plt.savefig("behavior_bar_chart.png")
plt.show()
```

```python
plt.figure(figsize=(6,6))
plt.pie(values, labels=labels, autopct='%1.1f%%')
plt.title("Behaviour Distribution")
plt.savefig("behavior_pie_chart.png")
plt.show()
```
