```python
from ultralytics import YOLO

model = YOLO("/content/drive/MyDrive/best.pt")

print("Model Loaded Successfully!")
```

```python
results = model.predict(
    source="/content/drive/MyDrive/train-2/val_batch0_pred.jpg",
    conf=0.25,
    save=True
)
```

```python
behavior_count = {}

for r in results:
    for box in r.boxes:
        cls = int(box.cls)
        label = model.names[cls]

        if label not in behavior_count:
            behavior_count[label] = 0

        behavior_count[label] += 1

print("Detected Behaviours:")
print(behavior_count)
```

```python
print("\n--- ALERT SYSTEM OUTPUT ---\n")

alerts_triggered = False

if behavior_count.get("sleep", 0) > 0:
    print("⚠ ALERT: Sleeping student detected!")
    alerts_triggered = True

if behavior_count.get("using_device", 0) > 0:
    print("⚠ ALERT: Mobile device usage detected!")
    alerts_triggered = True

if behavior_count.get("turn_head", 0) > 5:
    print("⚠ ALERT: Multiple students are distracted!")
    alerts_triggered = True

if not alerts_triggered:
    print(" No alerts. Classroom is normal.")
```

```python
with open("alert_log.txt", "w") as f:
    for key, value in behavior_count.items():
        f.write(f"{key}: {value}\n")

    f.write("\nAlert system executed successfully.\n")

print("Alert log saved!")
```

```python
behavior_count = {}

for r in results:
    for box in r.boxes:
        cls = int(box.cls)
        label = model.names[cls]

        if label not in behavior_count:
            behavior_count[label] = 0
        behavior_count[label] += 1


# Check alerts
alert_text = "NO ALERT - NORMAL CLASSROOM"

if behavior_count.get("sleep", 0) > 0:
    alert_text = "ALERT: Sleeping Detected!"
elif behavior_count.get("using_device", 0) > 0:
    alert_text = "ALERT: Mobile Usage Detected!"
```
