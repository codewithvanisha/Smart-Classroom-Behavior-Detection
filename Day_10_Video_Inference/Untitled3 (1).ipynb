```python
import cv2
import matplotlib.pyplot as plt

video_path = "/content/runs/detect/predict/8342243-uhd_3840_2160_25fps.avi"

cap = cv2.VideoCapture(video_path)

cap.set(cv2.CAP_PROP_POS_FRAMES, 100)

ret, frame = cap.read()

frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12,8))
plt.imshow(frame)
plt.axis("off")

cap.release()
```

```python
import cv2

video_path = "/content/runs/detect/predict/8342243-uhd_3840_2160_25fps.avi"

cap = cv2.VideoCapture(video_path)

for i, frame_no in enumerate([50, 150, 250]):
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_no)
    ret, frame = cap.read()
    if ret:
        cv2.imwrite(f"output_frame_{i+1}.jpg", frame)

cap.release()

print("Frames saved!")
```
